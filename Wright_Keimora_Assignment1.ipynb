#Part A
#The goal of this project is to predict whether a patient with diabetes will be readmitted to the hospital after being discharged. Hospital administrators, physicians, and other providers could use these predictions to identify patients who may need additional follow-up care or intervention before discharge. The prediction supports the risk of potential patient overlook and could benefit hospital resources. Success of this model would be measured by accurately identifying patients who are likely to be readmitted while minimizing missed high-risk patients and reducing overall readmission rates.

One way this problem could be improved would be to predict readmissions within different ranges. Another possible variation would be focusing only on older adults over age 65 to better understand readmission risk in that specific patient population.

#Part B
#Decision 1: Patient-Grouped Train/Test Split

An alternative approach would have been to randomly split all hospital visits into training and testing datasets without considering patient identifiers. We chose a patient-grouped train/test split so that visits from the same patient would not appear in both datasets. This helps create a more realistic evaluation because the model is tested on completely unseen patients.

One situation where this decision could hurt model performance is when there are patients with many visits that contain useful patterns.

Decision 2: ICD-9 Diagnosis Grouping

An alternative approach would have been to keep every ICD-9 diagnosis code as an individual category. We chose to group diagnosis codes into broader disease categories because there are thousands of possible codes, many of which occur very infrequently. Grouping reduces dimensionality and makes the data easier for the model to learn from.

This decision could negatively affect performance when specific diagnoses have a strong relationship with readmission risk.

#Part C
#I will compare Logistic Regression and Random Forest classification models. Logistic Regression provides a simple and interpretable baseline, while Random Forest can capture more complex relationships and interactions between variables.

I will report Precision, Recall, F1-Score, and ROC-AUC. Accuracy alone is not sufficient because the classes are imbalanced and a model could achieve high accuracy by predicting the majority class most of the time. Recall and F1-Score help evaluate how well the model identifies patients at risk of readmission, while ROC-AUC measures overall classification performance across different thresholds.

One risk I will monitor is class imbalance.

```python
#Part D
import pandas as pd

X_train = pd.read_csv("research/Predictive Modeling/X_train_scaled.csv")
X_test = pd.read_csv("research/Predictive Modeling/X_test_scaled.csv")
y_train = pd.read_csv("research/Predictive Modeling/y_train.csv").squeeze()
y_test = pd.read_csv("research/Predictive Modeling/y_test.csv").squeeze()
```

```python

import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)


#Logistic Regression

print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)
print("Training Labels Shape:", y_train.shape)
print("Testing Labels Shape:", y_test.shape)


print("\n==============================")
print("LOGISTIC REGRESSION MODEL")
print("==============================")

# Create model
log_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train model
log_model.fit(X_train, y_train)

# Generate predictions
log_predictions = log_model.predict(X_test)

# Generate probability predictions for ROC-AUC
log_probabilities = log_model.predict_proba(X_test)[:, 1]

# Display sample predictions
print("\nSample Predictions:")
print(log_predictions[:20])

# Evaluate Model
print("\nLogistic Regression Performance")

print("Accuracy:",
      round(accuracy_score(y_test, log_predictions), 4))

print("Precision:",
      round(precision_score(y_test, log_predictions), 4))

print("Recall:",
      round(recall_score(y_test, log_predictions), 4))

print("F1 Score:",
      round(f1_score(y_test, log_predictions), 4))

print("ROC-AUC:",
      round(roc_auc_score(y_test, log_probabilities), 4))

print("\nClassification Report:")
print(classification_report(y_test, log_predictions))

# ============================================================
# Model 2: Random Forest


print("\n==============================")
print("RANDOM FOREST MODEL")
print("==============================")

# Create model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Train model
rf_model.fit(X_train, y_train)

# Generate predictions
rf_predictions = rf_model.predict(X_test)

# Generate probability predictions
rf_probabilities = rf_model.predict_proba(X_test)[:, 1]

# Display sample predictions
print("\nSample Predictions:")
print(rf_predictions[:20])

# Evaluate Model
print("\nRandom Forest Performance")

print("Accuracy:",
      round(accuracy_score(y_test, rf_predictions), 4))

print("Precision:",
      round(precision_score(y_test, rf_predictions), 4))

print("Recall:",
      round(recall_score(y_test, rf_predictions), 4))

print("F1 Score:",
      round(f1_score(y_test, rf_predictions), 4))

print("ROC-AUC:",
      round(roc_auc_score(y_test, rf_probabilities), 4))

print("\nClassification Report:")
print(classification_report(y_test, rf_predictions))

# ============================================================
# Model Comparison Summary


comparison = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "F1 Score", "ROC-AUC"],
    "Logistic Regression": [
        accuracy_score(y_test, log_predictions),
        precision_score(y_test, log_predictions),
        recall_score(y_test, log_predictions),
        f1_score(y_test, log_predictions),
        roc_auc_score(y_test, log_probabilities)
    ],
    "Random Forest": [
        accuracy_score(y_test, rf_predictions),
        precision_score(y_test, rf_predictions),
        recall_score(y_test, rf_predictions),
        f1_score(y_test, rf_predictions),
        roc_auc_score(y_test, rf_probabilities)
    ]
})

print("\n==============================")
print("MODEL COMPARISON")
print("==============================")
print(comparison.round(4))
```

#Part E
#Overall, Logistic Regression performed better because it had higher accuracy, precision, and ROC-AUC. Although Random Forest had slightly higher recall and F1-score, the differences were very small. Based on these results, I would recommend Logistic Regression because it provided the strongest overall performance and better distinguished between readmitted and non-readmitted patients.

One surprising result was that Random Forest did not outperform Logistic Regression as expected. Despite being a more complex model, it produced lower precision and ROC-AUC scores, suggesting that a simpler model may work just as well for this dataset.#
